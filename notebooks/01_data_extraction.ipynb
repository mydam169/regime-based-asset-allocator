```python
import numpy as np
import pandas as pd

import yfinance as yf
from fredapi import Fred

import os
from dotenv import load_dotenv
load_dotenv()


import matplotlib.pyplot as plt
import seaborn as sns
```

# MacroEconomic Data

```python
fred_api_key = os.getenv('FRED_API_KEY')

start = '1990-01-01'
end = '2025-12-31'
```

```python
macro_variables = [
    # Real activity
    'INDPRO', 'W875RX1',
    # Labour market
    'UNRATE', 'ICSA', 'CCSA',
    # Inflation
    'CPIAUCSL',
    # Energy / commodities
    'DCOILWTICO',
    # Financial conditions
    'VIXCLS', 'BAA10Y', 'DBAA', 'DAAA', 'T10Y2Y',
    # Monetary policy
    'FEDFUNDS',
    # Sentiment & housing
    'UMCSENT', 'HOUST',
    # Money supply
    'M2SL',
    # Dollar index — three-way splice for full 1990 coverage:
    # TWEXMMTH (1973-present, major currencies, monthly)
    # TWEXB    (1995-2019, broad, weekly/monthly, discontinued)
    # DTWEXBGS (2006-present, broad goods+services, current)
    # Splice: TWEXMMTH for 1990-1994, TWEXB for 1995-2005,
    #         DTWEXBGS from 2006 onward.  Scale at each join.
    'TWEXMMTH', 'TWEXB', 'DTWEXBGS',
]
```

```python
fred = Fred(api_key=fred_api_key)
```

```python
series_dict = {}

for var in macro_variables:
    series_dict[var] = fred.get_series(var, observation_start=start,
                                        observation_end=end)

df_raw = pd.DataFrame(series_dict)
print(f'Raw data shape: {df_raw.shape}')
df_raw.head(10)
```

```python
df_raw.head(10)
```

```python
df_monthly = df_raw.resample('ME').last()
```

```python
df_monthly.head(10)
```

```python
df_monthly.isna().sum()
```

```python
# Forward-fill series released less frequently than monthly
for col in ['UNRATE', 'CPIAUCSL', 'CCSA',
            'TWEXMMTH', 'TWEXB', 'DTWEXBGS',
            'DBAA', 'DAAA']:
    if col in df_monthly.columns:
        df_monthly[col] = df_monthly[col].ffill()

# ── Three-way splice for full dollar index from 1990 ─────────────────
# Coverage by segment:
#   TWEXMMTH : 1973 – present  (major currencies, monthly)
#   TWEXB    : 1995 – 2019     (broad basket, discontinued)
#   DTWEXBGS : 2006 – present  (broad basket goods+services, current)
#
# Strategy:
#   1. Use DTWEXBGS from 2006 onward (most current, broadest)
#   2. Extend back to 1995 with TWEXB, scaled to match DTWEXBGS
#      at the Jan 2006 splice point
#   3. Extend back to 1990 with TWEXMMTH, scaled to match TWEXB
#      at the Jan 1995 splice point
# Both TWEXB and TWEXMMTH are on the same Federal Reserve H.10
# release as DTWEXBGS — same methodology, different baskets.

cols = ['TWEXMMTH', 'TWEXB', 'DTWEXBGS']
if all(c in df_monthly.columns for c in cols):

    splice1 = pd.Timestamp('2006-01-31')  # TWEXB → DTWEXBGS
    splice2 = pd.Timestamp('1995-01-31')  # TWEXMMTH → TWEXB

    # Scale TWEXB to match DTWEXBGS at splice1
    ov1 = df_monthly.loc[
        (df_monthly.index >= splice1) &
        df_monthly['TWEXB'].notna() &
        df_monthly['DTWEXBGS'].notna()
    ]
    scale1 = (ov1['DTWEXBGS'] / ov1['TWEXB']).median() if len(ov1) else 1.0

    # Scale TWEXMMTH to match scaled-TWEXB at splice2
    ov2 = df_monthly.loc[
        (df_monthly.index >= splice2) &
        df_monthly['TWEXMMTH'].notna() &
        df_monthly['TWEXB'].notna()
    ]
    scale2 = ((ov2['TWEXB'] * scale1) / ov2['TWEXMMTH']).median() if len(ov2) else 1.0

    seg_pre95  = df_monthly.loc[
        df_monthly.index < splice2, 'TWEXMMTH'
    ] * scale2
    seg_95_05  = df_monthly.loc[
        (df_monthly.index >= splice2) & (df_monthly.index < splice1), 'TWEXB'
    ] * scale1
    seg_post06 = df_monthly.loc[
        df_monthly.index >= splice1, 'DTWEXBGS'
    ]

    df_monthly['DOLLAR_INDEX'] = pd.concat([seg_pre95, seg_95_05, seg_post06])
    n_obs = df_monthly['DOLLAR_INDEX'].notna().sum()
    print(f'Dollar index spliced: {n_obs} observations '
          f'(scale1={scale1:.4f}, scale2={scale2:.4f})')

# ── Baa-Aaa spread ────────────────────────────────────────────────────
if 'DBAA' in df_monthly.columns and 'DAAA' in df_monthly.columns:
    df_monthly['BAA_AAA_SPREAD'] = df_monthly['DBAA'] - df_monthly['DAAA']
    print('Baa-Aaa spread computed')
```

```python
macro_var = {
    # Real activity
    'INDPRO':         'industrial_production',
    'W875RX1':        'real_person_income',

    # Labour market
    'UNRATE':         'unemployment_rate',
    'ICSA':           'initial_jobless_claims',
    'CCSA':           'continuing_claims',

    # Inflation
    'CPIAUCSL':       'cpi',

    # Energy / commodities
    'DCOILWTICO':     'oil_price',

    # Financial conditions
    'VIXCLS':         'vix',
    'BAA10Y':         'credit_spread',
    'BAA_AAA_SPREAD': 'baa_aaa_spread',
    'T10Y2Y':         'yield_curve_slope',

    # Monetary policy
    'FEDFUNDS':       'fed_funds_rate',

    # Sentiment & housing
    'UMCSENT':        'consumer_sentiment',
    'HOUST':          'housing_starts',

    # Money supply
    'M2SL':           'm2_money_supply',

    # Dollar index (TWEXB spliced with DTWEXBGS — see cell above)
    'DOLLAR_INDEX':   'dollar_index',
}

# Rename, drop raw series no longer needed, keep only mapped columns
df = df_monthly.rename(columns=macro_var)
drop_cols = [c for c in ['DBAA', 'DAAA', 'TWEXMMTH', 'TWEXB', 'DTWEXBGS']
             if c in df.columns]
df = df.drop(columns=drop_cols)
df = df[[c for c in macro_var.values() if c in df.columns]]
print(f'Final columns ({len(df.columns)}):')
print(list(df.columns))
df.head()
```

# Market Data

```python
tickers = ['VUSTX', 'VFINX']
```

```python
asset = yf.download(tickers=tickers, start='1990-01-01', end='2025-12-31', interval='1mo')['Close']
```

```python
asset.head()
```

```python
# data downloaded from gold.org
gld = pd.read_excel('../data/Gold_price_averages_in_a range_of_currencies_since_1978.xlsx',
                    sheet_name="Monthly_Avg", 
                    index_col=2,
                    header=5)
gld.index.name = "Date"
# selecting only USD price
gld = gld['USD']
gld.name = 'GLD'
gld.head()
```

```python
gld.index = gld.index.to_period('M').to_timestamp()
gld.head()
```

```python
asset_data = asset.join(gld, how="left").resample('ME').last()
```

```python
asset_data.head()
```

```python
market_var = {
    #Equity
    "VFINX": "index_fund",

    #Bonds
    "VUSTX": "treasury_fund",

    #Gold
    "GLD": "gold_fund"
}
```

```python
asset_data = asset_data.rename(columns=market_var)
```
